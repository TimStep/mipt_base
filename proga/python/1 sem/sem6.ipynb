```python
df = pd.DataFrame([[1, np.nan, np.nan, np.nan],
                   [np.nan, 1, np.nan, np.nan],
                   [np.nan, np.nan, 1, np.nan],
                   [np.nan, np.nan, np.nan, 1]],
                columns=list("ABCD"))

df[['A', 'B']] = df[['A', 'B']].fillna(2)
df.loc[0] = df.loc[0].fillna(1)
df['C'] = df['C'].fillna(2)
df = df.fillna(value = 1)
print(df)
```

<h1> 3

```python
m = 10
c = ''
for column in df.columns:
    if np.sum(df[column]) < m:
        c = column

df = df.drop([c], axis = 1)
print(df)
```
