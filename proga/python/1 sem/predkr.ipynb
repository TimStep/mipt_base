```python
def conv(s):
    n = 0
    for l in s:
        if l == 'v':
            n+=1
        if l == '<':
            n+=10
    return n

s = input()
num = s.split('.')[::-1]
print(num)

res = 0

for i in range(len(num)): res += conv(num[i])*(60**i)
print(res)
```

```python
def conv_back(n):
    if n == 0:
        return '@'
    sticks = n//5
    rocks = n%5
    res = rocks*'.' + sticks*'|'
    return res

n= int(input())
num = []

while n>0:
    num.append(n%20)
    n = n // 20

num = num[::-1]
print(num)
num = [conv_back(i) for i in num]
print(' '.join(num))
```
