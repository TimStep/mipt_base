<h1> 5

```python
states = df.State.unique()
#print(states)
TDCA = []

for state in states:
    df2 = df[df['State'] == state]
    TDCA.append(np.mean(df2['Total day calls']))
#print(ATDC)

df2 = pd.DataFrame(TDCA, index = states)
print(df2)
```

<h1> 6

```python
A = np.sum(df['Total day calls']) / len(df)
print(A)

df2 = df2[df2[0] > A]
print(df2)
```

<h1> 7+8

```python
dict = {}
dict['States'] = states
dict['Total day calls average'] = TDCA

TECA = []
DorE = []
for i in range(len(states)):
    df2 = df[df['State'] == states[i]]
    TECA.append(np.mean(df2['Total eve calls']))
    if TDCA[i] > TECA[i]:
        DorE.append(True)
    else:
        DorE.append(False)

dict['Total eve calls average'] = TECA
dict['Day > Evening'] = DorE
df3 = pd.DataFrame(dict)
print(df3)
```
