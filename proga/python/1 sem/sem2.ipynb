```python
n = int(input())

res = {}
for i in range(n):
    movie = input()
    if movie not in res:
        res[movie] = 1
    else:
        res[movie] +=1

#print(res)
res = dict(sorted(res.items(), key = lambda item: -1*item[1]))
#print(res)

for movie in res:
    print(movie + ' ' + str(res[movie]))
```

```python
text = input()

res = {}
for i in range(len(text)):
    letter = text[i]
    if letter == " ": continue
    if letter not in res:
        res[letter] = 1
    else:
        res[letter] += 1

res = {i: res[i] for i in sorted(list(res.keys()))}
#print(res)

for letter in res:
    print(letter + ' ' + str(res[letter]))
```

```python
def parents(tree, name):
    return list(tree[name])

def children(tree, name):
    ans = []
    for child in tree:
        if name in tree[child]: ans.append(child)
    return ans

def grandparents(tree, name):
    gparnts = []
    parnts = parents(tree, name)
    parnts_check = parents(tree, name)
    for parnt in parnts:
        if parnt not in tree: parnts_check.remove(parnt)
    if parnts_check == []: return [0]
    for parnt in parnts: gparnts.append(parents(tree, parnt))
    return gparnts

def grandchildren(tree, name):
    gchldrn = []
    for chld in children(tree, name): gchldrn.append(children(tree, chld))
    return gchldrn

def siblings(tree, name):
    sibls = []
    prnts1 = parents(tree, name)
    for rel in tree:
        if rel == name: continue
        prnts2 = parents(tree, rel)
        for prnt in prnts1:
            if prnt in prnts2:
                sibls.append(rel)
                break
    return sibls

tr = {

'C': {'A', 'B'},

'F': {'C'},
    
'D': {'C'}
}

print(*parents(tr, 'C'))
print(*children(tr, 'C'))
print(*grandparents(tr, 'C'))
print(*grandparents(tr, 'F'))
print(*grandchildren(tr, 'A'))
print(*siblings(tr, 'F'))
```
