<h1> RegEx

```python
import re
```

```python
s = input()

result = re.findall('([A-Z][a-z]+)([A-Z][a-z]+)', s)
print(result)
```

```python
s = input()

if re.match('[АВЕКМНОРСТУХ]{1}\d{3}[АВЕКМНОРСТУХ]{2}\d+', s):
    print('Private')
elif re.match('[АВЕКМНОРСТУХ]{2}\d+', s):
    print('Taxi')
else:
    print('Fail')
```

```python
s = input()

s = re.sub('[0-1][0 - 9]:[0-5][0-9]:[0-5][0-9]', 'TBS', s)
s = re.sub('2[0 - 3]:[0-5][0-9]:[0-5][0-9]', 'TBS', s)
s = re.sub('[0-1][0 - 9]:[0-5][0-9]', 'TBS', s)
s = re.sub('2[0 - 3]:[0-5][0-9]', 'TBS', s)

print(s)
```

```python
s = input()

s = re.split('( |\.)', s)
#print(s)
for w in range(len(s)):
    try:
        s[w] = str(int(s[w])**3)
    except:
        continue

#print(s)
print(''.join(s))
```

<h1> Stats

```python
import statistics as stats
import csv
```

```python
dat = list(map(int, input().split()))

print(stats.mean(dat), stats.median(dat), stats.mode(dat))
```

```python
dat1 = list(map(int, input().split()))
dat2 = list(map(int, input().split()))

f = open('statistics_res.csv', 'w', encoding = 'UTF8', newline = '')
writer = csv.writer(f)

writer.writerow(['Последовательность 1'])
writer.writerow(['Среднее'] + [str(stats.mean(dat1))])
writer.writerow(['Медиана'] + [str(stats.median(dat1))])
writer.writerow(['Мода'] + [str(stats.mode(dat1))])
writer.writerow(['Стандратное отклонение'] + [str(stats.stdev(dat1))])
writer.writerow(['Дисперсия'] + [str(stats.variance(dat1))])

writer.writerow(['Последовательность 2'])
writer.writerow(['Среднее'] + [str(stats.mean(dat2))])
writer.writerow(['Медиана'] + [str(stats.median(dat2))])
writer.writerow(['Мода'] + [str(stats.mode(dat2))])
writer.writerow(['Стандратное отклонение'] + [str(stats.stdev(dat2))])
writer.writerow(['Дисперсия'] + [str(stats.variance(dat2))])

f.close()
```
