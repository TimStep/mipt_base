```python
f = open('chat.txt', 'r')
lines = f.readlines()
lines = [line.rstrip() for line in lines]
users = []
newlines = []
for line in lines:
    line = line.split('-')
    line[0] = line[0].rstrip()
    user = line[0]
    line[1] = line[1].rstrip()
    phrase = line[1]
    
    newlines.append(line)
    
    if user not in users: users.append(user)
    #print(line)

#print(newlines)
#print(users)
#print(lines)

phrases = [[] for i in range(len(users))]
#print(phrases)

for pair in newlines:
    phrases[users.index(pair[0])].append(pair[1])

#print(phrases)

for user in users:
    print(user + ' said ' + str(len(phrases[users.index(user)])) + ' phrases')

total_symbols = 0
for phrase in phrases:
    for message in phrase:
        total_symbols += len(message)

print('Average message length:', total_symbols/len(users))
```

```python
res = open('results.txt', 'w')
for user in users:
    res.write(user + ' said ' + str(len(phrases[users.index(user)])) + ' phrases\n')
res.write('Average message length: ' + str(total_symbols/len(users)))
f.close()
res.close()
```
