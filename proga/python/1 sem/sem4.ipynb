```python
import numpy as np
import matplotlib.pyplot as plt
import statistics as stats
```

```python
x = np.linspace(0.0, 5.0, 10)
y = 2 * x - 5
eps = np.random.randn(y.shape[0])
y = y + eps * 0.2

k = (stats.mean([x[i]*y[i] for i in range(len(x))]) - stats.mean(x)*stats.mean(y)) / (stats.mean([i**2 for i in x]) - stats.mean(x)**2)
b = stats.mean(y) - k*stats.mean(x)

plt.figure(figsize = (10,8))
plt.scatter(x, y)
plt.plot(x, k*x + b)

'''
k, b = np.polyfit(x, y, 1)
plt.plot(x, k*x + b, color = 'orange')
'''
```

```python
A = np.zeros((3, 5))
#print(A)
for i in range(A.shape[0]): A[i] = np.random.randn(A.shape[1])

print('Матрица:')
print(A)

print('Наименьшее наибольших по столбцам:')
m = max(A.T[0])
for column in A.T:
    localmax = max(column)
    if localmax < m: m = localmax
print(m)

print('Умноженная на себя транспонированную:')
print(A @ A.T)

print('Стандартное отклонение')
print('По строкам:')
for row in A: print(row.std())
print('По столбцам:')
for column in A.T: print(column.std())

print('По всей матрице:')
'''
values = []
for row in A: values.extend(row)
values = np.array(values)
#print(values)
print(values.std())
'''
print(A.std())

sizes = np.linspace(10, 1010, 101)
#print(sizes)
s = []
for n in sizes:
    M = np.zeros((int(n), int(n)))
    for i in range(M.shape[0]): M[i] = np.random.randn(M.shape[1])
    s.append(M.std())
plt.scatter(sizes, s)
```
