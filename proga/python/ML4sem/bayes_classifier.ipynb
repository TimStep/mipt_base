```python
import numpy as np
from sklearn import datasets
iris = datasets.load_iris()

X, y = iris.data, iris.target
```

```python
data = np.hstack((X, y.reshape(150, 1)))
np.random.shuffle(data)
train, test = data[0:80], data[81:101]
trainX, trainY, testX, testY = np.delete(train, 4, 1), train[:, 4], np.delete(test, 4, 1), test[:, 4]
trainX.shape, trainY.shape, testX.shape, testY.shape
```

```python
class MinMaxScaler():

    def __init__(self):

        self.mins = None
        self.maxs = None

    def fit(self, data):

        self.maxs = data.max(axis=0)
        self.mins = data.min(axis=0)

        return None
    
    def transform(self, data):

        transformed = []
        for i in range(data.shape[1]):
            
            std = (data[:, i] - self.mins[i]) / (self.maxs[i] - self.mins[i])
            transformed.append(std)
        
        return np.array(transformed).T
    
    def fit_transform(self, data):

        self.fit(data)
        return self.transform(data)
```

```python
data = np.array([[-1, 2], [-0.5, 6], [0, 10], [1, 18]])
scaler = MinMaxScaler()
print(scaler.fit_transform(data))
print(scaler.maxs)
print(scaler.mins)
print(scaler.transform(np.array([[2, 2]])))
```
