```python
import pandas as pd 
from sklearn import datasets
from sklearn.utils import shuffle
import sklearn.metrics as metrics
iris = datasets.load_iris()
```

```python
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['target'] = iris.target
df.head()
```

```python
df = shuffle(df)
df.reset_index(inplace=True, drop=True)
df.head()
```

```python
class KNN():

    def __init__(self, df, k, sep, len):
        
        self.k = k
        train, test = df.iloc[0:sep], df.iloc[sep+1:len+1]
        self.X_train, self.X_test, self.y_train, self.y_test = train.drop(["target"], axis=1), test.drop(["target"], axis=1), train[["target"]], test[["target"]]

    def fit_and_predict(self):
        
        y_pred = self.y_test.copy()

        for i, row in self.X_test.iterrows():
            temp_train = self.X_train - row
            temp_train = temp_train.pow(2)
            temp_train["sum"] = temp_train.sum(axis=1)
            temp_train = temp_train.sort_values("sum")
            m = self.y_train.loc[temp_train.index[:self.k]].mode()["target"][0]
            y_pred.loc[i] = m

        return y_pred
    
    def score(self):

        preds = self.fit_and_predict()
        acc = (preds == self.y_test)
        return acc.mean()
```

```python
iris_classifier = KNN(df, 5, 80, 100)
iris_classifier.score()
```
