#**6. Fine-tuning**

image.png

Пример тонкой настройки (fine-tuning) модели VGG16 на датасете CIFAR10 с заморозкой части слоев в Keras:

- Замораживаем первые 3 блока сверток (15 слоев), которые отвечают за базовые паттерны, и размораживаем последние 2 блока для адаптации под CIFAR10
- Изменен размер входа (32x32x3 вместо 224x224x3)
- Добавлены новые полносвязные слои поверх базовой модели

````python
from tensorflow.keras.applications import VGG16
from tensorflow.keras import layers, models, optimizers
from tensorflow.keras.datasets import cifar10
from tensorflow.keras.utils import to_categorical

# 1. Загрузка данных
(x_train, y_train), (x_test, y_test) = cifar10.load_data()
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

# 2. Загрузка предобученной модели VGG16 (без верхних слоев)
base_model = VGG16(
    weights='imagenet',
    include_top=False,
    input_shape=(32, 32, 3)
)

# 3. Заморозка начальных слоев (первые 3 блока)
for layer in base_model.layers:
    if 'block1' in layer.name or 'block2' in layer.name or 'block3' in layer.name:
        layer.trainable = False  # Замораживаем
    else:
        layer.trainable = True    # Размораживаем последние 2 блока

# 4. Добавление кастомных слоев
model = models.Sequential([
    base_model,
    layers.Flatten(),
    layers.Dense(256, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(10, activation='softmax')
])

print(model.summary())

print("Чтобы проверить, какие слои заморожены:")
for i, layer in enumerate(base_model.layers):
    print(f"Layer {i}: {layer.name} - Trainable: {layer.trainable}")

# 5. Компиляция модели
model.compile(
    optimizer=optimizers.Adam(learning_rate=1e-4),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# 6. Обучение
#his = model.fit(x_train, y_train, epochs=5, batch_size=32, validation_data=(x_test, y_test))

#model.evaluate(x_test, y_test)

#plt.plot(his.history['loss'])
#plt.plot(his.history['val_loss'])
#plt.show()
````

#**7. Keras Tuner**

Основные изменения:

1. Создана функция `model_builder` с возможностями настройки:
   - Количество фильтров в сверточных слоях (32-128 и 64-256)
   - Размер ядра свертки (3x3 или 5x5)
   - Количество нейронов в плотном слое (64-256)
   - Размер Dropout (0.1-0.5)
   - Learning rate (0.01, 0.001 или 0.0001)

2. Использован тюнер Hyperband для эффективного поиска параметров
3. Добавлено раннее прекращение обучения (EarlyStopping)

````python
import os
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '2'

import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import mnist
from tensorflow import keras
from tensorflow.keras.layers import Dense, Flatten, Dropout, Conv2D, MaxPooling2D
import keras_tuner as kt

(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Стандартизация и преобразование данных
x_train = x_train / 255
x_test = x_test / 255

y_train_cat = keras.utils.to_categorical(y_train, 10)
y_test_cat = keras.utils.to_categorical(y_test, 10)

x_train = np.expand_dims(x_train, axis=3)
x_test = np.expand_dims(x_test, axis=3)

def model_builder(hp):
    model = keras.Sequential()

    # Подбор параметров для первого сверточного слоя
    model.add(Conv2D(
        filters=hp.Int('conv1_filters', min_value=32, max_value=128, step=32),
        kernel_size=hp.Choice('conv1_kernel', values=[3, 5]),
        padding='same',
        activation='relu',
        input_shape=(28, 28, 1)
    ))
    model.add(MaxPooling2D((2, 2), strides=2))

    # Подбор параметров для второго сверточного слоя
    model.add(Conv2D(
        filters=hp.Int('conv2_filters', min_value=64, max_value=256, step=64),
        kernel_size=hp.Choice('conv2_kernel', values=[3, 5]),
        padding='same',
        activation='relu'
    ))
    model.add(MaxPooling2D((2, 2), strides=2))

    model.add(Flatten())

    # Подбор параметров для плотного слоя
    model.add(Dense(
        units=hp.Int('dense_units', min_value=64, max_value=256, step=64),
        activation='relu'
    ))

    # Подбор Dropout
    model.add(Dropout(
        rate=hp.Float('dropout_rate', min_value=0.1, max_value=0.5, step=0.1)
    ))

    model.add(Dense(10, activation='softmax'))

    # Подбор learning rate
    hp_learning_rate = hp.Choice('learning_rate', values=[1e-2, 1e-3, 1e-4])

    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=hp_learning_rate),
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )

    return model

# Инициализация тюнера
tuner = kt.Hyperband(
    model_builder,
    objective='val_accuracy',
    max_epochs=10,
    factor=3,
    directory='my_tuner',
    project_name='mnist_tuning'
)

# Поиск гиперпараметров
tuner.search(
    x_train,
    y_train_cat,
    epochs=5,
    validation_split=0.2,
    batch_size=32,
    callbacks=[keras.callbacks.EarlyStopping(monitor='val_loss', patience=2)]
)

# Получение лучшей модели
best_model = tuner.get_best_models(num_models=1)[0]

# Обучение лучшей модели
history = best_model.fit(
    x_train,
    y_train_cat,
    epochs=10,
    batch_size=32,
    validation_split=0.2
)

# Оценка модели
best_model.evaluate(x_test, y_test_cat)

# Вывод результатов
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title('Model Loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation'], loc='upper left')
plt.show()
````

#**8. ResNet**

image.png

image.png

Основные изменения:
1. Заменена Sequential модель на Functional API
2. Добавлены остаточные блоки с skip-connections
3. Использована комбинация Conv2D + BatchNorm + ReLU
4. Добавлены проекционные shortcut-соединения при изменении размерности
5. Заменен Flatten + Dense на GlobalAveragePooling

Архитектура теперь включает:
- Начальный сверточный слой (32 фильтра)
- Два остаточных блока с 32 фильтрами
- Два остаточных блока с 64 фильтрами (с уменьшением размера в 2 раза)
- Глобальный усредняющий пулинг вместо Flatten

````python
import os
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '2'

import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import mnist
from tensorflow import keras
from tensorflow.keras.layers import Input, Conv2D, BatchNormalization, Activation, Add, GlobalAveragePooling2D, Dense
from tensorflow.keras.models import Model

(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Стандартизация и преобразование данных
x_train = x_train / 255.0
x_test = x_test / 255.0

y_train_cat = keras.utils.to_categorical(y_train, 10)
y_test_cat = keras.utils.to_categorical(y_test, 10)

x_train = np.expand_dims(x_train, axis=-1)
x_test = np.expand_dims(x_test, axis=-1)

# Функция остаточного блока
def residual_block(input_tensor, filters, stride=1):
    shortcut = input_tensor

    x = Conv2D(filters, (3, 3), strides=stride, padding='same')(input_tensor)
    x = BatchNormalization()(x)
    x = Activation('relu')(x)

    x = Conv2D(filters, (3, 3), strides=1, padding='same')(x)
    x = BatchNormalization()(x)

    if stride != 1 or shortcut.shape[-1] != filters:
        shortcut = Conv2D(filters, (1, 1), strides=stride, padding='same')(shortcut)
        shortcut = BatchNormalization()(shortcut)

    x = Add()([x, shortcut])
    x = Activation('relu')(x)
    return x

# Создание модели ResNet-like
def build_resnet(input_shape=(28, 28, 1), num_classes=10):
    inputs = Input(shape=input_shape)

    # Начальный слой
    x = Conv2D(32, (3, 3), padding='same')(inputs)
    x = BatchNormalization()(x)
    x = Activation('relu')(x)

    # Последовательность остаточных блоков
    x = residual_block(x, filters=32, stride=1)
    x = residual_block(x, filters=32, stride=1)
    x = residual_block(x, filters=64, stride=2)
    x = residual_block(x, filters=64, stride=1)

    # Финальные слои
    x = GlobalAveragePooling2D()(x)
    outputs = Dense(num_classes, activation='softmax')(x)

    return Model(inputs, outputs)

model = build_resnet()
model.summary()

model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

history = model.fit(x_train, y_train_cat,
                    batch_size=32,
                    epochs=10,
                    validation_split=0.2)

model.evaluate(x_test, y_test_cat)

# Визуализация обучения
plt.plot(history.history['loss'], label='train_loss')
plt.plot(history.history['val_loss'], label='val_loss')
plt.legend()
plt.show()
````
